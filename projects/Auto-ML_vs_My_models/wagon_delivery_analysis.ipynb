### Поработаем с фичами типа datetime + небольшой Feature Engineering
Попробуем из дат вытащить информацию, которую мы сможем использовать в дальнейшем
 
Например: прошедшее количество суток с последних ремонтов, в какое время года проходит перевозка

*Важное уточнение, если использовать модель в реальных случаях, мы должны учитывать только факты, которые мы знаем на момент начала рейса*

```python
def get_season_vectorized(dates):
    months = dates.dt.month
    return np.select(
        [months.isin([12, 1, 2]), 
         months.isin([3, 4, 5]),
         months.isin([6, 7, 8]),
         months.isin([9, 10, 11])],
        ['ЗИМА', 'ВЕСНА', 'ЛЕТО', 'ОСЕНЬ'],
        default='Unknown'
    )

def get_month_vectorized_ru(dates):
    # Возвращает названия месяцев на русском
    month_names_ru = {
        1: 'Январь', 2: 'Февраль', 3: 'Март', 4: 'Апрель',
        5: 'Май', 6: 'Июнь', 7: 'Июль', 8: 'Август',
        9: 'Сентябрь', 10: 'Октябрь', 11: 'Ноябрь', 12: 'Декабрь'
    }
    return dates.dt.month.map(month_names_ru)

def get_weekday_vectorized_ru(dates):
    # Возвращает названия дней недели
    weekday_names_ru = {
        0: 'Понедельник', 1: 'Вторник', 2: 'Среда',
        3: 'Четверг', 4: 'Пятница', 5: 'Суббота', 6: 'Воскресенье'
    }
    return dates.dt.dayofweek.map(weekday_names_ru)

def is_weekend_vectorized_ru(dates, format='Да/Нет'):
    # Проверяет, является ли день выходным 
    is_weekend = dates.dt.dayofweek.isin([5, 6])  # 5 - суббота, 6 - воскресенье
    
    if format == 'Да/Нет':
        return np.where(is_weekend, 'Да', 'Нет')
    elif format == 'Выходной/Рабочий':
        return np.where(is_weekend, 'Выходной', 'Рабочий')
    else:
        return is_weekend
    
    
# Определим сезонные признаки
info_cargo_imp['Сезон'] = get_season_vectorized(info_cargo_imp['Дата начала'])
info_cargo_imp['Месяц'] = get_month_vectorized_ru(info_cargo_imp['Дата начала'])
info_cargo_imp['День недели'] = get_weekday_vectorized_ru(info_cargo_imp['Дата начала'])
info_cargo_imp['Выходной'] = is_weekend_vectorized_ru(info_cargo_imp['Дата начала'], format='Выходной/Рабочий')

# Некоторые строки datetime переведём в удобный формат
info_cargo_imp['Количество суток до планового ремонта'] = (info_cargo_imp[' Дата планового ремонта'] - info_cargo_imp['Дата начала']).dt.days
info_cargo_imp['Количество суток до конца службы вагона'] = (info_cargo_imp['Утв. дата продления срока службы'] - info_cargo_imp['Дата начала']).dt.days
info_cargo_imp['Отведённое количество суток на доставку по нормативам'] = (info_cargo_imp['Нормативный срок доставки'] - info_cargo_imp['Дата начала']).dt.days

# Также нам нужно знать, будет ли проводится ремонт вовремя рейса, основываясь на нормативном сроке доставке 
info_cargo_imp['Плановый ремонт во время рейса'] = (info_cargo_imp['Дата следующего планового ремонта'] >
                                                    info_cargo_imp['Дата и время начала рейса']) & (
                                                               info_cargo_imp['Дата следующего планового ремонта'] <=
                                                               info_cargo_imp['Нормативный срок доставки'])
# Будет ли проведена операция во время рейса
info_cargo_imp['Планируется ли операция во время рейса'] = (info_cargo_imp['Дата и время операции'] >
                                                    info_cargo_imp['Дата и время начала рейса']) & (
                                                               info_cargo_imp['Дата и время операции'] <=
                                                               info_cargo_imp['Нормативный срок доставки'])


# Удаляем ненужные колонки - datetime после обработки колонки
datetime_columns = info_cargo_imp.select_dtypes(include = ['datetime']).columns
info_cargo_imp = info_cargo_imp.drop(columns = datetime_columns)
```

```python
# Так как мы не обработали пропущенные значения в датах, они скорей всего должны появится в наших новых данных
null_index = info_cargo_imp.columns[info_cargo_imp.isnull().any()].tolist()
print(null_index)
info_cargo_imp[null_index] = info_cargo_imp[null_index].fillna(info_cargo_imp[null_index].mean())
```

Согласно ГОСТ 33211-2014 и правилам РЖД вес вагонов зимой часто ниже летнего на 2-5%. Проверим, действительно ли это так

```python
season_weight = info_cargo_imp[info_cargo_imp != 0 ].groupby(['Сезон'], as_index = False)['Вес груза (кг)'].mean()

sns.barplot(x = season_weight ['Сезон'],
            y = season_weight ['Вес груза (кг)'])
plt.ylim(60000, 65000)
plt.show()

diff = season_weight[season_weight['Сезон'] == 'ЗИМА']['Вес груза (кг)'][0] / season_weight[(season_weight['Сезон'] == 'ЛЕТО')]['Вес груза (кг)']

print(f'Различие массы груза Летом в сравнение с Зимой составляет {(diff[1] - 1) * 100}%')
print(f'Граница 2%, которая должна быть около {season_weight['Вес груза (кг)'][0] * 1.02}, наш же тест показал {season_weight['Вес груза (кг)'][1]}')
```

Давайте сделаем небольшой статистический тест, является ли это статистически значимой разницей?

$H0 ~ \mu = 64893$ - Вес груза Летом превышает зимний на 2%

$H1 ~ \mu < 64893$ - Веc груза Летом не привышает зимний на 2%

```python
sample = info_cargo_imp[info_cargo_imp['Вес груза (кг)'] != 0 ]['Сезон'][info_cargo_imp['Сезон'] == 'ОСЕНЬ'].size
sample_mean = info_cargo_imp[info_cargo_imp['Вес груза (кг)'] != 0 ]['Вес груза (кг)'][info_cargo_imp['Сезон'] == 'ОСЕНЬ'].mean()
mean_H0 = season_weight['Вес груза (кг)'][0] * 1.02
sample_std = info_cargo_imp[info_cargo_imp['Вес груза (кг)'] != 0 ]['Вес груза (кг)'][info_cargo_imp['Сезон'] == 'ОСЕНЬ'].std()
SE = (sample_std / np.sqrt(sample))

# n > 100 мы можем утверждать что работаем со стандартным отклонением ГС, используем Z тест 
z_stat = (sample_mean - mean_H0) / SE
p_value = stats.norm.cdf(z_stat)  # Для H₁: μ < μ₀

print(f"z-статистика = {z_stat:.3f}, p-value = {p_value:.6f}")

alpha = 0.05
if p_value < alpha:
    print(f"Отвергаем H₀: среднее выборки < {mean_H0} (p < 0.05)")
else:
    print("Нет оснований отвергнуть H₀")
    
ci_low = sample_mean - stats.norm.ppf(0.975) * SE
ci_high = sample_mean + stats.norm.ppf(0.975) * SE
print(f"95% ДИ: [{ci_low:.2f}, {ci_high:.2f}]")
```

На самом деле это было очевидно просто по графику, что в действительности Летом меньше перевозят чем Зимой, но сделать такой стат тест будет не лишним.

```python
# -----------------------------------------------------------------------------------------------------------------------
# Для Catboost

X = info_cargo_imp.drop(columns=['Срок доставки факт'])
y = info_cargo_imp['Срок доставки факт']

categorical_features_indices = info_cargo_imp.select_dtypes(include=['object']).columns.tolist()
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

train_pool = cb.Pool(data = X_train, 
    label = y_train, 
    cat_features = categorical_features_indices)

test_pool = cb.Pool(data = X_test,  
                    cat_features = categorical_features_indices)
# -----------------------------------------------------------------------------------------------------------------------
# Для XBboost

X_xgb = info_cargo_imp.drop(columns=['Срок доставки факт'])
y_xgb = info_cargo_imp['Срок доставки факт']

X_encoded = X_xgb.copy()  # Создаем копию, чтобы не изменять исходный DataFrame
for col in X_encoded.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    X_encoded[col] = le.fit_transform(X_encoded[col])

X_train_en, X_test_en, y_train_en, y_test_en = train_test_split(X_encoded, y_xgb, test_size=0.2, random_state=0)

# -----------------------------------------------------------------------------------------------------------------------
# LightGBM

X_lgb = info_cargo_imp.drop(columns=['Срок доставки факт'])
y_lgb = info_cargo_imp['Срок доставки факт']

for col in categorical_features_indices:
      X_lgb[col] = X_lgb[col].astype('category')

X_train_lgb, X_test_lgb, y_train_lgb, y_test_lgb = train_test_split(X_lgb, y_lgb, test_size=0.2, random_state=0)
```

## Подбор параметров

```python
# -----------------------------------------------------------------------------------------------------------------------
# CatBoost - подбор параметров

param_distributions = { 
    'learning_rate': stats.uniform(0.05, 0.5),
    'depth': stats.randint(4, 11),
    'l2_leaf_reg': stats.uniform(1, 10)
}

random_search_cb = RandomizedSearchCV(
    estimator=cb.CatBoostRegressor(iterations = 1000, cat_features = categorical_features_indices, random_seed=42, verbose=0, loss_function='RMSE', task_type='GPU'), 
    param_distributions=param_distributions,
    n_iter=10,  # Количество случайных комбинаций для перебора
    scoring='neg_root_mean_squared_error', # Выберите подходящую метрику
    cv=3, # Кросс-валидация
    random_state=42
)

random_search_cb.fit(X_train, y_train, eval_set=[(X_test, y_test)], early_stopping_rounds=50, verbose=False)

print("Лучшие параметры для catboost:", random_search_cb.best_params_)
print("Лучшие параметры для catboost:", random_search_cb.best_score_)


param_grid = {
     'depth': [9,10,11],
     'learning_rate': [0.05, 0.1, 0.15, 0.2],
     'l2_leaf_reg': [7,8,9],
 }
 
grid_search = GridSearchCV(cb.CatBoostRegressor(iterations=1000, cat_features = categorical_features_indices , loss_function='RMSE', nan_mode='Min', task_type='GPU'), param_grid, cv=3, scoring='neg_root_mean_squared_error', n_jobs=1)
grid_search.fit(X_train, y_train, eval_set=[(X_test, y_test)], early_stopping_rounds=50, verbose=False)
 
print("Лучшие параметры Catboost при 1000 итераций:", grid_search.best_params_)
print("Лучшие параметры Catboost при 1000 итераций:", grid_search.best_score_)

# -----------------------------------------------------------------------------------------------------------------------
# Xgboost - подбор параметров

param_distributions_xgb = { 
    'n_estimators': stats.randint(100, 10000),
    'learning_rate': stats.uniform(0.005, 0.1),
    'max_depth': stats.randint(4, 14),
    'gamma': stats.uniform(0.1, 0.9)
}

random_search_xgb = RandomizedSearchCV(
    estimator= xgb.XGBRegressor(objective='reg:squarederror', subsample=0.8,
                             colsample_bytree=0.8), 
    param_distributions=param_distributions_xgb,
    n_iter=50,  # Количество случайных комбинаций для перебора
    scoring='neg_root_mean_squared_error', # Выберите подходящую метрику
    cv=3, # Кросс-валидация
    random_state=42
)

random_search_xgb.fit(X_train_en, y_train_en, eval_set=[(X_test_en, y_test_en)], verbose=0)

print("Лучшие параметры для XGboost:", random_search_xgb.best_params_)
print("Лучший результат для XGboost:", random_search_xgb.best_score_)
# -----------------------------------------------------------------------------------------------------------------------
# LightGBM - подбор параметров

param_distributions_lgb = {
    'n_estimators': stats.randint(100, 10000),
    'learning_rate': stats.uniform(0.001, 0.1),
    'max_depth': stats.randint(3, 30),
    'num_leaves': stats.randint(20, 100),
    'min_child_samples': stats.randint(5, 50),
    'subsample': stats.uniform(0.6, 0.4),
    'colsample_bytree': stats.uniform(0.6, 0.4), 
    'reg_alpha': stats.uniform(0, 1),
    'reg_lambda': stats.uniform(0, 1)
}

random_search_lgb = RandomizedSearchCV(
    estimator= lgb.LGBMRegressor(
    objective='regression',
    metric='rmse',
    random_state=42),
    param_distributions=param_distributions_lgb,
    n_iter=300, 
    scoring='neg_root_mean_squared_error', 
    cv=3,       
    random_state=42,
    n_jobs=-1,   
    verbose=1    
)

random_search_lgb.fit(X_train_lgb, y_train_lgb,
                  eval_set=[(X_test_lgb, y_test_lgb)],
                  eval_metric='rmse',
                  callbacks=[lgb.early_stopping(stopping_rounds=200, verbose=False)])  # Ранняя остановка

print("Лучшие параметры для LightGBM:", random_search_lgb.best_params_)
print("Лучший результат для LightGBM", random_search_lgb.best_score_)
```

## Обучение моделей

```python
# -----------------------------------------------------------------------------------------------------------------------
# Catboost - Обучение

start_time_cb = time.time()
model_cb = cb.CatBoostRegressor(iterations = 3000 , 
                             l2_leaf_reg = 7,
                          depth = 10, 
                          learning_rate = 0.01, 
                          loss_function = 'RMSE', 
                          nan_mode = 'Min', 
                          random_seed = 314,
                          verbose = 0,
                             task_type='GPU')
model_cb.fit(train_pool)
end_time_cb = time.time() 
training_time_cb = end_time_cb - start_time_cb

# -----------------------------------------------------------------------------------------------------------------------
# Xgboost - Обучение

start_time_xgb = time.time()
model_xgb = xgb.XGBRegressor(objective='reg:squarederror', 
                             n_estimators=7500,
                             learning_rate=0.001,
                             gamma = 0.2,
                             max_depth=11,
                             subsample=0.8,
                             colsample_bytree=0.8,
                             random_state=42)

model_xgb.fit(
    X_train_en, y_train_en,
    eval_set=[(X_test_en, y_test_en)],  # Для ранней остановки
    verbose=False)
end_time_xgb = time.time() 
training_time_xgb = end_time_xgb - start_time_xgb

# -----------------------------------------------------------------------------------------------------------------------
# LightGBM - Обучение

start_time_lgb = time.time()
model_lgb = lgb.LGBMRegressor(
    objective='regression',      
    metric='rmse',               
    n_estimators= 2000,           
    learning_rate=0.013,          
    num_leaves=73,               
    max_depth=27,                
    subsample=0.75,               
    colsample_bytree=0.9,        
    reg_alpha=0.5,               
    reg_lambda=0.6,              
    random_state=42,             
)

model_lgb.fit(X_train_lgb, y_train_lgb,
          eval_set=[(X_test_lgb, y_test_lgb)],  # Валидационный набор
          eval_metric='rmse',          # Метрика для оценки (root mean squared error)
          callbacks=[lgb.early_stopping(stopping_rounds=1000)])

end_time_lgb = time.time()
training_time_lgb = end_time_lgb - start_time_lgb
```

### Тестирование моделей 

```python
results = []

# -----------------------------------------------------------------------------------------------------------------------
# Catboost - Тестирование

model_cb.save_model('Best_models/catboost_model.cbm')

test_predict = model_cb.predict(test_pool)
MSE_cb = mean_squared_error(y_test, test_predict)
MAE_cb = mean_absolute_error(y_test, test_predict)
RMSE_cb = root_mean_squared_error(y_test, test_predict)
R_square_cb = r2_score(y_test, test_predict)
results.append(['CatBoost', MSE_cb, MAE_cb, RMSE_cb, R_square_cb,training_time_cb ])

# -----------------------------------------------------------------------------------------------------------------------
# Xgboost - Тестирование

model_xgb.save_model('Best_models/xgboost_model.ubj')

test_predict_XGB = model_xgb.predict(X_test_en)
MSE_XGB = mean_squared_error(y_test_en, test_predict_XGB)
MAE_XGB = mean_absolute_error(y_test_en, test_predict_XGB)
RMSE_XGB = root_mean_squared_error(y_test_en, test_predict_XGB)
R_square_XGB = r2_score(y_test_en, test_predict_XGB)
results.append(['XGBoost', MSE_XGB, MAE_XGB, RMSE_XGB,R_square_XGB, training_time_xgb])

# -----------------------------------------------------------------------------------------------------------------------
# LightGBM - Тестирование

joblib.dump(model_lgb, 'Best_models/lightgbm_model.cbm')

test_predict_lgb = model_lgb.predict(X_test_lgb)
MSE_lgb = mean_squared_error(y_test_lgb, test_predict_lgb)
MAE_lgb = mean_absolute_error(y_test_lgb, test_predict_lgb)
RMSE_lgb = root_mean_squared_error(y_test_lgb, test_predict_lgb)
R_square_lgb = r2_score(y_test_lgb, test_predict_lgb)
results.append(['LightGBM', MSE_lgb, MAE_lgb, RMSE_lgb, R_square_lgb, training_time_lgb])

# -----------------------------------------------------------------------------------------------------------------------
results_df = pd.DataFrame(results, columns=['Модель', 'MSE', 'MAE', 'RMSE','R^2' ,'Время обучения (сек)']).sort_values('MAE', ascending=True)
print("\nТаблица результатов:")
results_df
```

   - MAE = 1.61 дня (лучший результат среди всех моделей)
